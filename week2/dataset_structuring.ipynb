```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

sold = pd.read_csv('../week1/combined_sold.csv')
listings = pd.read_csv('../week1/combined_listings.csv')
```

```python
print(sold.shape)
print(sold.columns.tolist())
print(sold.dtypes)
print(sold.head())
print(sold.info())
```

```python
null_counts = sold.isnull().sum()
null_pct = (sold.isnull().sum() / len(sold)) * 100

missing_report = pd.DataFrame({
    'null_count': null_counts,
    'null_pct': null_pct.round(2)
}).sort_values('null_pct', ascending=False)

print(missing_report.to_string())
```

```python
high_missing = missing_report[missing_report['null_pct'] > 90]
print("Columns with >90% missing values:")
print(high_missing)
```
